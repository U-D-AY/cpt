```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = np.array(range(1, 100000, 1000))
y = np.log(x)
plt.plot(x, y)
plt.grid()
plt.show()
```

```python
x = np.array(range(0, 100, 5))
y = np.sin(x)
plt.figure(figsize = (10,6), dpi=100)
plt.plot(x, y, color="red", lw = 2, marker = "*", markerfacecolor = "cyan", markersize=20)
plt.grid()
plt.show()
```

```python
x = np.array(range(0, 100, 5))
y = np.cos(x)
plt.figure(figsize = (10,6), dpi=100)
plt.plot(x, y, color="aqua", ls = 'solid', lw = 2, marker = "^", markerfacecolor = "purple", markersize=10, markeredgecolor="black")
plt.grid()
plt.show()
```

```python
x = np.array(range(1, 100, 5))
y = (np.cos(x)+x**3-x**2-2)
plt.figure(figsize = (10,6), dpi=100)
plt.plot(x, y, color="aqua", ls = ":", lw = 2, marker = "o", markerfacecolor = "lime", markersize=10, markeredgecolor="yellow")
plt.title('Line plot Referal', fontsize=16, va='top', ha='left', fontweight='bold', color="magenta")
plt.xlabel("Year of Working", fontsize=12)
plt.ylabel("Porfit in Million", fontsize=12)
plt.gca().set_facecolor('lemonchiffon')
plt.gcf().set_facecolor('cyan')

plt.grid()
plt.show()
```
