- Simple nested lambda string combining with seperator operation for n attributes.

```python
o = (lambda a: lambda b: lambda c, d: a + ' ' + b + ' ' + c + ' ' + d)
print(o("Hello")("my")("name", "Uday"))
```

- Given attributes: a, b, c, d are hi, !, students, !!!

- Write a nexted lambda to concatnate the abcd, by converting a and c to uppercase
- Output: 
- Hi!Students!!!

```python
print((lambda a: lambda b: lambda c, d: a+b+c+d)('Hi')('!')('Students',"!!!"))
```

```python
print((lambda a: lambda b: lambda c, d: " ".join(map(lambda x: x[::-1],(a,b,c,d))))('Hi')('!')('Students',"!!!"))
```

```python
# above with list comprehention

print(' '.join([x[::-1] for x in ("Hi", '!', "Students", "!!!")]))
```
